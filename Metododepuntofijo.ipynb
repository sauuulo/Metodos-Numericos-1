Método de punto fijo.

Queremos resolver la ecuación $x^3 + 4x^2 - 10 = 0$ en el intervalo $[1, 2]$.

El teorema del punto fijo nos asegura que si $g(x) \in [a, b]$ para todo $x \in [a, b]$ y $\vert{}g'(x)\vert{} < 1$ en ese intervalo, existe un único punto fijo $p \in [a, b]$ tal que $g(p) = p$, el cual corresponde a una raíz de $f(p) = 0$.  

```python

import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
    return x**3+4*x**2-10
```

```python
xx=np.linspace(1,2)
plt.plot(xx,f(xx))
plt.title("Gráfica de la función f(x) = x³ + 4x² - 10")
plt.axhline(y=0, color='green')
plt.axvline(x=0, color='red')
```

```python
def g(x):
    return 0.5*np.sqrt(10-x**3)
```

```python
g(1.5)
```

```python
def pfijo(p0,i):
   lista=[]
   error=[]

   for i in range(10):
      p1=g(p0)
      lista.append(p1)
      e=np.abs(p1-p0)
      error.append(e)
      p0=p1
   return lista, error
```

```python
lista, error =pfijo(1.5,15)
```

A continuación se verá la aproximación obtenida en cada iteración.

```python
for i, valor in enumerate(lista):
    print(f"Iteración {i + 1}: {valor}")
```
